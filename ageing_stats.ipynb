# Dataset statistics — 5 s segments

```python
from pathlib import Path

import pandas as pd

PARTS = sorted(Path(".").glob("features_261_5s_part*of2.parquet"))
df = pd.concat([pd.read_parquet(p) for p in PARTS], ignore_index=True)

print([p.name for p in PARTS])
print(df.shape)
assert df["utterance_id"].is_unique
```

## Overview

```python
pd.Series({
    "utterances": len(df),
    "videos": df["video_id"].nunique(),
    "speakers": df["VoxCeleb_ID"].nunique(),
})
```

## Speaker age

```python
age = df["speaker_age"]
n_multi_age = (df.groupby("VoxCeleb_ID")["speaker_age"].nunique() > 1).sum()

pd.Series({
    "min": age.min(),
    "max": age.max(),
    "mean": round(age.mean(), 1),
    "std": round(age.std(), 1),
    "median": age.median(),
    "speakers with more than one age": n_multi_age,
})
```

## Subsets

```python
speakers = df.drop_duplicates("VoxCeleb_ID")


def summarize(d, s):
    return pd.Series({
        "utterances": len(d),
        "speakers": d["VoxCeleb_ID"].nunique(),
        "female": (s["gender"] == "female").sum(),
        "male": (s["gender"] == "male").sum(),
        "videos": d["video_id"].nunique(),
    })


table = pd.DataFrame({
    name: summarize(d, speakers[speakers["subset"] == name])
    for name, d in df.groupby("subset", sort=False)
}).T
table.loc["total"] = summarize(df, speakers)
table
```

```python
n_both = (df.groupby("VoxCeleb_ID")["subset"].nunique() > 1).sum()
print(f"speakers in both subsets: {n_both}")
```
